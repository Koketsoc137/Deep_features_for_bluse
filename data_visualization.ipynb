# BLUSE Data Exploration

This notebook is set up for the BLUSE dataset on your desktop. The structure contains:
- a data folder with HDF5 files
- a features folder with extracted NumPy feature arrays
- an examples folder with sample images

Run the cells below to inspect the data and create visualizations.

```python
from pathlib import Path
import os

base_path = Path.home() / 'Desktop' / 'BLUSE'
print('BLUSE path:', base_path)
print('Exists:', base_path.exists())

for folder in ['data', 'features', 'examples']:
    p = base_path / folder
    print(f'\n{folder}:', p)
    print('exists:', p.exists())
    if p.exists():
        items = list(p.iterdir())
        print('items:', len(items))
        for item in items[:10]:
            print(' -', item.name)
        if len(items) > 10:
            print(' ...')
```

```python
import numpy as np
from pathlib import Path

base_path = Path.home() / 'Desktop' / 'BLUSE'
feature_path = base_path / 'features' / 'mk_sample_resnet18_patch_features.npy'

if feature_path.exists():
    X = np.load(feature_path, mmap_mode='r')
    print('Feature array shape:', X.shape)
    print('dtype:', X.dtype)
    print('mean:', np.mean(X), 'std:', np.std(X))
    print('first row shape:', X[0].shape)
else:
    print('Feature file not found:', feature_path)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

base_path = Path.home() / 'Desktop' / 'BLUSE'
feature_path = base_path / 'features' / 'mk_sample_resnet18_patch_features.npy'

if feature_path.exists():
    X = np.load(feature_path, mmap_mode='r')
    # Normalization suggested in the README
    l1 = X[..., :64] / np.linalg.norm(X[..., :64], axis=-1, keepdims=True)
    l2 = X[..., 64:] / np.linalg.norm(X[..., 64:], axis=-1, keepdims=True)
    Xn = np.concatenate([l1, l2], axis=-1)
    features = np.concatenate([Xn.mean(axis=1), Xn.max(axis=1)], axis=-1)

    plt.figure(figsize=(10, 5))
    sns.histplot(features.mean(axis=1), bins=40, kde=True)
    plt.title('Distribution of mean feature values')
    plt.xlabel('Mean feature value')
    plt.ylabel('Count')
    plt.tight_layout()
    plt.show()
else:
    print('Feature file not found. Add the dataset or update the file name.')
```
